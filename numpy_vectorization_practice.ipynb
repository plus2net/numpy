```python
import numpy as np
import time

# Benchmark 1,000,000 additions
size = 1000000
a = np.arange(size)
b = np.arange(size)

t0 = time.time()
c = a + b
t1 = time.time()
print(f'Vectorized addition time: {t1 - t0:.6f} seconds')
```
