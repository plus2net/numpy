**Objectives:**
1. Benchmark execution time between Python `for` loops and NumPy array addition.
2. Execute zero-loop conditional filtering using `np.where()`.

4. Test 2D matrix broadcasting rules.

### Step 1: Element-wise Addition Benchmark (1,000,000 items)

```python
import numpy as np
import time

size = 1000000
py_list_a = list(range(size))
py_list_b = list(range(size))
np_arr_a = np.arange(size)
np_arr_b = np.arange(size)

# Python loop
t0 = time.time()
py_res = [x + y for x, y in zip(py_list_a, py_list_b)]
t_py = time.time() - t0

# NumPy vectorization
t0 = time.time()
np_res = np_arr_a + np_arr_b
t_np = time.time() - t0

print(f'Python Loop Time: {t_py:.5f} sec')
print(f'NumPy Vectorized Time: {t_np:.5f} sec')
print(f'Speedup: {t_py / t_np:.1f}x faster!')
```

### Step 2: Zero-Loop Conditional Masking (`np.where`)

```python
matrix = np.random.randn(500, 500)
filtered_matrix = np.where(matrix < 0, 0.0, matrix)
print('Negative values successfully replaced without loops!')
```

### Step 3: Math ufunc Benchmark (`np.sin` vs `math.sin`)

```python
import math

arr_data = np.random.rand(500000)
list_data = arr_data.tolist()

t0 = time.time()
sin_py = [math.sin(x) for x in list_data]
t_py_sin = time.time() - t0

t0 = time.time()
sin_np = np.sin(arr_data)
t_np_sin = time.time() - t0

print(f'Ufunc speedup: {t_py_sin / t_np_sin:.1f}x faster!')
```

### Step 4: Exercise Verification

```python
arr = np.random.randn(500000)
res = np.where(arr > 0, arr**2, 0.0)

assert np.all(res >= 0)
print('Vectorization exercise verification passed!')
```
