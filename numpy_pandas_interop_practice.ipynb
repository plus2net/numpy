### 1. NumPy to Pandas Conversion

```python
import numpy as np
import pandas as pd

arr = np.arange(6).reshape(2, 3)
df = pd.DataFrame(arr, columns=['A', 'B', 'C'])
print("DataFrame:\n", df)
```

### 2. Pandas to NumPy and Memory Sharing

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'X': [10, 20], 'Y': [30, 40]})
arr_view = df.to_numpy()
print("Extracted array:\n", arr_view)
print("Shares memory:", np.shares_memory(df['X'], arr_view))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np
import pandas as pd

# Test 1: DataFrame creation shape and index
data = np.array([[85, 90, 95], [78, 82, 88], [92, 96, 94]])
df = pd.DataFrame(data, index=['Alice', 'Bob', 'Charlie'], columns=['Math', 'Science', 'English'])
assert df.shape == (3, 3)
assert df.index[0] == 'Alice'

# Test 2: to_numpy extraction and dtype promotion with NaN
df_nan = pd.DataFrame({'A': [1, 2], 'B': [3.5, np.nan]})
arr_nan = df_nan.to_numpy()
assert arr_nan.dtype == np.float64

# Test 3: Memory isolation with copy=True
df_mem = pd.DataFrame({'X': [10, 20]})
arr_c = df_mem.to_numpy(copy=True)
assert np.shares_memory(df_mem['X'], arr_c) == False

print("All numpy & pandas interop unit assertions passed successfully! 100% verified.")
```
