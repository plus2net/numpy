### 1. Object Identity, Type & Value

```python
greeting = "plus2net"
items = [1, 2, 3]

print("id(greeting):", id(greeting), "type:", type(greeting))
print("id(items):   ", id(items), "type:", type(items))
```

### 2. In-Place Mutation vs. Reassignment

```python
# Mutable list: in-place update
nums = [10, 20, 30]
addr_before = id(nums)
nums.append(40)
print("List id unchanged:", id(nums) == addr_before)

# Immutable integer: reassignment creates new object
val = 100
addr_val_before = id(val)
val += 1
print("Int id changed:   ", id(val) != addr_val_before)
```

### 3. Nested Immutability Trap

```python
record = ("Alice", [88, 92])
print("Before:", record)

# The tuple is immutable, but the inner list can mutate!
record[1].append(96)
print("After: ", record)
```

### 4. Avoiding Mutable Default Function Arguments

```python
# Safe idiom using None sentinel
def add_item_safe(item, container=None):
    if container is None:
        container = []
    container.append(item)
    return container

print("Call 1:", add_item_safe("Alpha"))
print("Call 2:", add_item_safe("Beta"))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: List mutability
lst = [1, 2]
old_id = id(lst)
lst += [3]
assert id(lst) == old_id

# Test 2: Tuple immutability
t = (1, 2)
t_id = id(t)
t += (3,)
assert id(t) != t_id

# Test 3: Unhashable list
try:
    hash([1, 2])
    assert False
except TypeError:
    pass

print("All mutable vs. immutable assertions successfully verified! 100% passed.")
```
