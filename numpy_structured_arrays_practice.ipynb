## Exercise 1: Parsing Raw Binary Financial Ticks
Parse a packed binary byte stream directly into a structured array using `np.frombuffer`.

```python
import numpy as np

# Define financial tick schema
tick_dtype = np.dtype([
    ('ticker', 'S4'),
    ('timestamp', 'f8'),
    ('price', 'f8'),
    ('volume', 'i4')
])

# Binary buffer
raw_binary_stream = (
    b"AAPL" + np.float64(1625097600.0).tobytes() + np.float64(145.50).tobytes() + np.int32(500).tobytes() +
    b"GOOG" + np.float64(1625097601.0).tobytes() + np.float64(2500.25).tobytes() + np.int32(120).tobytes()
)

ticks = np.frombuffer(raw_binary_stream, dtype=tick_dtype)

print("Parsed Tickers:", ticks['ticker'])
print("Prices:        ", ticks['price'])
print("Total Volume:  ", ticks['volume'].sum())

# Self-validating test assertions
assert ticks['ticker'][0] == b"AAPL"
assert ticks['volume'].sum() == 620
```

## Exercise 2: Record Arrays (`np.recarray`) & Dot Notation
Create a recarray view to access fields directly using object attribute access.

```python
log_dtype = np.dtype([('sensor_id', 'u2'), ('voltage', 'f4'), ('error', '?')])

rec_logs = np.array([
    (101, 3.28, False),
    (102, 4.12, True),
    (103, 3.31, False)
], dtype=log_dtype).view(np.recarray)

print("Voltages via attribute:", rec_logs.voltage)
print("Error Count:           ", rec_logs.error.sum())

assert rec_logs.voltage[0] == 3.28
assert rec_logs.error.sum() == 1
```

## Exercise 3: Relational Boolean Queries on Binary Network PCAP Data
Filter packet capture logs based on multi-field boolean conditions without database overhead.

```python
pcap_dtype = np.dtype([
    ('src_ip', 'S15'),
    ('dst_port', 'u2'),
    ('size', 'u4'),
    ('latency', 'f4')
])

packets = np.array([
    (b"192.168.1.10", 80, 1500, 12.4),
    (b"10.0.0.55", 443, 512, 125.8),
    (b"192.168.1.12", 80, 1420, 8.1),
    (b"10.0.0.55", 22, 64, 2.1)
], dtype=pcap_dtype)

query_mask = (packets['dst_port'] == 80) & (packets['latency'] < 15.0)
filtered_packets = packets[query_mask]

print("Filtered Packet Count:", len(filtered_packets))
print("Average Size (bytes) :", filtered_packets['size'].mean())

assert len(filtered_packets) == 2
assert filtered_packets['size'].mean() == 1460.0
```
