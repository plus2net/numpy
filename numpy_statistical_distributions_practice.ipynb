### 1. Normal and Uniform Distributions

```python
import numpy as np

np.random.seed(42)
norm_samples = np.random.normal(loc=50.0, scale=5.0, size=100)
print("Normal sample mean:", np.mean(norm_samples))

uni_samples = np.random.uniform(low=10.0, high=20.0, size=5)
print("Uniform samples:", uni_samples)
```

### 2. Binomial and Poisson Distributions

```python
import numpy as np

np.random.seed(42)
bin_samples = np.random.binomial(n=10, p=0.5, size=5)
print("Binomial successes:", bin_samples)

pois_samples = np.random.poisson(lam=3.0, size=5)
print("Poisson events:", pois_samples)
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Normal distribution mean approx
np.random.seed(42)
ns = np.random.normal(loc=50.0, scale=1.0, size=5000)
assert round(np.mean(ns), 0) == 50.0

# Test 2: Uniform range bounds
us = np.random.uniform(low=5.0, high=10.0, size=100)
assert np.all(us >= 5.0) and np.all(us < 10.0)

# Test 3: Binomial shape
bs = np.random.binomial(10, 0.5, size=(2, 3))
assert bs.shape == (2, 3)

print("All numpy statistical distributions unit assertions passed successfully! 100% verified.")
```
