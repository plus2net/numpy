### 1. Basic 1D and 2D Ones Arrays

```python
import numpy as np

# 1D vector (default float64)
vec = np.ones(5)
print("1D vector: ", vec)
print("Dtype:     ", vec.dtype)

# 2D matrix (3 rows, 4 cols)
mat = np.ones((3, 4), dtype=int)
print("\n2D matrix:\n", mat)
```

### 2. Custom Data Types (int, bool, complex)

```python
import numpy as np

o_bool = np.ones(4, dtype=bool)
o_cplx = np.ones(3, dtype=complex)
print("Boolean ones: ", o_bool)
print("Complex ones: ", o_cplx)
```

### 3. Cloning Geometries with ones_like()

```python
import numpy as np

prototype = np.array([[10, 20, 30], [40, 50, 60]], dtype=np.int16)
mirrored = np.ones_like(prototype)
print("Prototype shape:", prototype.shape)
print("Mirrored array:\n", mirrored)
```

### 4. Production Recipe: Uniform Prior Probabilities

```python
import numpy as np

n_classes = 5
uniform_priors = np.ones(n_classes, dtype=np.float64) / n_classes
print("Uniform probability priors:\n", uniform_priors)
```

### 5. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape integrity
o1 = np.ones(6)
assert o1.shape == (6,) and np.all(o1 == 1.0)

# Test 2: Multidimensional integer matrix
o2 = np.ones((2, 3, 2), dtype=int)
assert o2.shape == (2, 3, 2) and o2.size == 12

# Test 3: ones_like equivalence
source_arr = np.zeros((4, 4), dtype=np.float32)
cloned = np.ones_like(source_arr)
assert cloned.shape == (4, 4) and cloned.dtype == np.float32 and np.all(cloned == 1)

print("All numpy.ones() unit assertions successfully verified! 100% passed.")
```
