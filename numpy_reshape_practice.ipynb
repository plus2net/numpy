### Learning Objectives:
1. Reshape 1D arrays into 2D matrices.
2. Automatically infer dimensions using `-1`.
3. Compare C-style and Fortran-style memory orders.
4. Run a self-verifying assertion test suite.

### 1. Basic Reshaping and Inference

```python
import numpy as np

ar = np.arange(12)
print("Reshaped 3x4:\n", ar.reshape(3, 4))
print("\nInferred 4x3:\n", ar.reshape(4, -1))
```

### 2. Memory Order (C vs Fortran)

```python
import numpy as np

ar = np.arange(6)
print("C order:\n", ar.reshape(2, 3, order='C'))
print("\nF order:\n", ar.reshape(2, 3, order='F'))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape matching
ar = np.arange(12)
assert ar.reshape(3, 4).shape == (3, 4)

# Test 2: Inference matching
assert np.array_equal(ar.reshape(4, -1), ar.reshape(4, 3))

# Test 3: Ravel and flatten dimensions
mat = np.array([[1, 2], [3, 4]])
assert mat.ravel().ndim == 1 and mat.flatten().ndim == 1

print("All numpy.reshape() unit assertions passed successfully! 100% verified.")
```
