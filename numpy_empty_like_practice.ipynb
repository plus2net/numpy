### 1. Basic Prototype Cloning

```python
import numpy as np

ar = np.array([4, 7, 9])
ar_empty = np.empty_like(ar)
print("Shape:", ar_empty.shape)
print("Dtype:", ar_empty.dtype)
```

### 2. Dtype and Shape Overrides

```python
import numpy as np

ar = np.array([4, 7, 9])
ar_float = np.empty_like(ar, dtype=float)
ar_matrix = np.empty_like(ar, shape=(2, 3))

print("Float dtype array dtype:", ar_float.dtype)
print("Overridden shape matrix shape:\n", ar_matrix.shape)
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape and dtype cloning integrity
source = np.array([1, 2, 3], dtype=np.int32)
cloned = np.empty_like(source)
assert cloned.shape == source.shape and cloned.dtype == source.dtype

# Test 2: Dtype override
f_cloned = np.empty_like(source, dtype=np.float64)
assert f_cloned.dtype == np.float64

# Test 3: Shape override
s_cloned = np.empty_like(source, shape=(4, 2))
assert s_cloned.shape == (4, 2)

print("All numpy.empty_like() unit assertions passed successfully! 100% verified.")
```
