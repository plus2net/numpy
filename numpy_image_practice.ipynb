# NumPy Image Array Manipulation Notebook

This practice notebook accompanies the [NumPy Image Array Manipulation Tutorial on plus2net](https://www.plus2net.com/python/numpy-image-manipulation.php).

**Objectives:**
1. Create and inspect 3D RGB image array tensors `(Height, Width, Channels)`.
2. Crop image regions using 2D/3D array slice indexing.
3. Implement pixel color inversion (`255 - img`) and boolean thresholding (`np.where`).
4. Extract single RGB channels and calculate ITU-R BT.601 weighted grayscale arrays.
5. Adjust dynamic contrast and brightness using `np.clip()` to prevent `uint8` overflow.

### Step 1: Initialize Synthetic 3D RGB Image Tensor

```python
import numpy as np

# Synthetic 300x300 RGB image (uint8)
img_tensor = np.random.randint(0, 256, size=(300, 300, 3), dtype=np.uint8)
print('Image Tensor Shape:', img_tensor.shape)
print('Data Type:', img_tensor.dtype)
```

### Step 2: Region Cropping via Slicing

```python
# Crop central 100x100 pixel area
cropped = img_tensor[100:200, 100:200, :]
print('Cropped Tensor Shape:', cropped.shape)
```

### Step 3: Color Inversion & Binary Thresholding

```python
# Invert image colors
inverted = 255 - img_tensor

# Boolean thresholding for bright pixels (> 200)
thresholded = np.where(img_tensor > 200, np.uint8(255), np.uint8(0))
print('Thresholding completed successfully.')
```

### Step 4: Channel Splitting & Grayscale Conversion

```python
red_ch = img_tensor[:, :, 0]
print('Red Channel Matrix Shape:', red_ch.shape)

# ITU-R BT.601 Grayscale
gray = (0.2989 * img_tensor[:, :, 0] + 0.5870 * img_tensor[:, :, 1] + 0.1140 * img_tensor[:, :, 2]).astype(np.uint8)
print('Grayscale Matrix Shape:', gray.shape)
```

### Step 5: Linear Contrast & Brightness Adjustment

```python
img_float = img_tensor.astype(np.float32)
adjusted = np.clip(img_float * 1.25 + 40, 0, 255).astype(np.uint8)
print('Adjusted Image Tensor Min/Max:', adjusted.min(), adjusted.max())
```

### Step 6: Exercise Verification

```python
img = np.zeros((100, 100, 3), dtype=np.uint8)
img[25:75, 25:75, 0] = 255

assert img[50, 50, 0] == 255
print('Exercise verification passed! Average intensity:', np.mean(img))
```
