# NumPy min(): Axes, Masks, and Minimum Value Aggregation

**Tutorial & Documentation:** [https://www.plus2net.com/python/numpy-agg-min.php](https://www.plus2net.com/python/numpy-agg-min.php)

Welcome to the plus2net interactive companion notebook for `numpy.min()`. Run each cell sequentially in Google Colab or your local Jupyter environment.

### Learning Objectives:
1. Calculate overall, row (axis=1), and column (axis=0) minimum values.
2. Preserve dimensions with `keepdims=True` for broadcast normalization.
3. Apply Boolean `where` inclusion masks with safety `initial` ceilings.
4. Manage missing values with `np.nanmin()`.
5. Distinguish reduction `min()` from element-wise `minimum()` and index finder `argmin()`.
6. Run a self-verifying assertion suite.

### 1. Overall, Column, and Row Reductions

```python
import numpy as np

a = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
print("Sample Matrix:\n", a)

# Reductions across different axes
print("\nOverall minimum:          ", a.min())
print("Column minimums (axis=0): ", a.min(axis=0))
print("Row minimums (axis=1):    ", a.min(axis=1))
```

### 2. Output Buffer Reuse and keepdims

```python
import numpy as np

a = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])

# Preallocated output buffer
col_buffer = np.zeros(3, dtype=int)
a.min(axis=0, out=col_buffer)
print("Stored in preallocated out buffer:", col_buffer)

# Retaining reduced dimensions with keepdims=True
row_mins_kd = a.min(axis=1, keepdims=True)
print("\nRow minimums with keepdims=True shape:", row_mins_kd.shape)
print("Distance above row minimum:\n", a - row_mins_kd)
```

### 3. Boolean Masks (where) and Initial Comparison Ceilings

```python
import numpy as np

a = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
mask = np.array([True, False, True])

# Include columns 0 and 2 only, competing against initial ceiling
masked_mins = a.min(axis=1, where=mask, initial=10)
print("Masked row minimums (initial=10):", masked_mins)
```

### 4. Handling NaN Values with nanmin

```python
import numpy as np

data_nan = np.array([4.5, np.nan, 1.8, 9.2])

print("Standard np.min() propagates NaN:", np.min(data_nan))
print("np.nanmin() ignores NaN:          ", np.nanmin(data_nan))
```

### 5. Element-Wise Minimum vs. Reduction vs. Index Finder

```python
import numpy as np

x = np.array([10, 20, 30])
y = np.array([15, 12, 35])

# Element-wise comparison
print("np.minimum(x, y):", np.minimum(x, y))

# argmin finds index of the minimum
scores = np.array([88, 42, 95, 42, 70])
print("First minimum index with argmin():", np.argmin(scores))
print("All tie indices:                   ", np.where(scores == scores.min())[0])
```

### 6. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Overall and directional reductions
arr = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
assert arr.min() == 2
assert np.array_equal(arr.min(axis=0), np.array([6, 2, 2]))
assert np.array_equal(arr.min(axis=1), np.array([2, 2, 2]))

# Test 2: keepdims geometry
kd = arr.min(axis=1, keepdims=True)
assert kd.shape == (3, 1)

# Test 3: Masked where with initial ceiling
m = arr.min(axis=1, where=np.array([True, False, True]), initial=10)
assert np.array_equal(m, np.array([2, 2, 6]))

# Test 4: nanmin finite extraction
n_arr = np.array([12.0, np.nan, 3.5, 8.0])
assert np.isnan(np.min(n_arr))
assert np.nanmin(n_arr) == 3.5

# Test 5: Pairwise minimum
p_min = np.minimum(np.array([5, 10]), np.array([8, 3]))
assert np.array_equal(p_min, np.array([5, 3]))

print("All numpy.min() unit assertions passed successfully! 100% verified.")
```
