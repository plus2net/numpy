### 1. Basic Countdown Recursion

```python
def countdown(n):
    if n < 0:
        print("Blastoff!")
        return
    print(n, end=", ")
    countdown(n - 1)

countdown(5)
```

### 2. Factorial Function

```python
def factorial(n):
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)

print("5! =", factorial(5))
print("7! =", factorial(7))
```

### 3. Memoized Fibonacci with LRU Cache

```python
import functools

@functools.lru_cache(maxsize=None)
def fib(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

print("Fibonacci(30):", fib(30))
print("Fibonacci(50):", fib(50))
```

### 4. Recursive List Flattening

```python
def flatten(nested):
    out = []
    for x in nested:
        if isinstance(x, list):
            out.extend(flatten(x))
        else:
            out.append(x)
    return out

matrix = [1, [2, [3, 4], 5], 6]
print("Flattened:", flatten(matrix))
```

### 5. Self-Checking Unit Verification Suite

```python
# Test 1: Factorial
assert factorial(0) == 1
assert factorial(4) == 24

# Test 2: Fibonacci
assert fib(0) == 0
assert fib(1) == 1
assert fib(10) == 55

# Test 3: Flattening
assert flatten([10, [20, [30]]]) == [10, 20, 30]

print("All recursion unit assertions successfully verified! 100% passed.")
```
