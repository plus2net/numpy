### 1. Generating Single Random Integers

```python
import numpy as np

np.random.seed(42)

# 1. Range [0, 4) -> {0, 1, 2, 3}
val_low_only = np.random.randint(4)
print("Random integer [0, 4):", val_low_only)

# 2. Range [4, 10) -> {4, 5, 6, 7, 8, 9}
val_range = np.random.randint(4, 10)
print("Random integer [4, 10):", val_range)
```

### 2. Generating 1D and 2D Arrays with size

```python
import numpy as np

np.random.seed(100)

# 1D vector of 5 integers from [4, 10)
vec = np.random.randint(4, 10, size=5)
print("1D Vector (5 items):\n", vec)

# 2D matrix (3 rows x 2 cols)
mat = np.random.randint(4, high=10, size=(3, 2))
print("\n2D Matrix (3 rows x 2 cols):\n", mat)
print("Shape:", mat.shape)
```

### 3. Specifying Integer Precision with dtype

```python
import numpy as np

np.random.seed(20)

# 3x3 matrix with 16-bit integer precision
arr16 = np.random.randint(4, high=7, size=(3, 3), dtype='int16')
print("3x3 Matrix with dtype='int16':\n", arr16)
print("Dtype:    ", arr16.dtype)
print("Item size:", arr16.itemsize, "bytes")
```

### 4. Practical Recipes: RGB Color Generator & Dice Roll Simulation

```python
import numpy as np

# Recipe A: Generate a random RGB color (values 0-255)
color_rgb = tuple(np.random.randint(0, 256, size=3))
print("Random RGB Color Tuple:", color_rgb)

# Recipe B: Simulate rolling two fair dice 1,000 times
np.random.seed(42)
dice_rolls = np.random.randint(1, 7, size=(1000, 2))
totals = dice_rolls.sum(axis=1)

print("\nDice Rolls Simulation (1000 rolls):")
print("Empirical mean total (expected ~7.0):", np.mean(totals))
print("Minimum total:", totals.min(), "Maximum total:", totals.max())
```

### 5. Modern Generator Alternative: default_rng().integers()

```python
import numpy as np

# Recommended modern NumPy API
rng = np.random.default_rng(seed=123)

# endpoint=True includes the high bound (6)
modern_rolls = rng.integers(low=1, high=6, endpoint=True, size=10)
print("Modern Generator die rolls [1, 6]:", modern_rolls)
```

### 6. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Bound verification on large sample
test_sample = np.random.randint(5, 15, size=10000)
assert np.all(test_sample >= 5) and np.all(test_sample < 15)

# Test 2: Shape and dimension integrity
mat_test = np.random.randint(0, 10, size=(4, 5, 2))
assert mat_test.shape == (4, 5, 2) and mat_test.ndim == 3

# Test 3: Data type enforcement
u8_arr = np.random.randint(0, 256, size=10, dtype=np.uint8)
assert u8_arr.dtype == np.uint8 and u8_arr.itemsize == 1

# Test 4: Verify low >= high raises ValueError
try:
    np.random.randint(10, 5)
    assert False, "Failed to raise ValueError on low >= high"
except ValueError:
    pass

print("All numpy.random.randint() practice assertions passed successfully!")
```
