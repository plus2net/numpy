**Objectives:**
1. Understand array stride steps in memory.
2. Construct zero-copy sliding window views using `sliding_window_view`.
3. Compute rolling window averages without memory overhead.

```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

arr = np.array([10, 20, 30, 40, 50, 60, 70])
view = sliding_window_view(arr, window_shape=3)
rolling_avg = view.mean(axis=-1)
print('Rolling averages:', rolling_avg)
```
