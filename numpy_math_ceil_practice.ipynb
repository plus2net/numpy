### Learning Objectives:
1. Compute ceiling values for floating-point arrays.
2. Utilize preallocated `out` buffers for performance optimization.
3. Apply boolean `where` masks for selective operations.
4. Run a self-verifying assertion test suite.

### 1. Basic ceil() Execution

```python
import numpy as np

ar = np.array([1.57, 2.09, 2.79, -3.41, -3.62])
print("Ceil result:", np.ceil(ar))
```

### 2. out Buffer and where Mask

```python
import numpy as np

ar = np.array([1.57, 2.09, 2.79, -3.41, -3.62])
ar_out = np.empty_like(ar)
np.ceil(ar, out=ar_out)
print(
    "Preallocated out buffer:", ar_out
)

ar_where = np.array([True, False, True, False, True])
print("Masked ceil:", np.ceil(ar, where=ar_where))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Basic ceiling equality
ar = np.array([1.57, 2.09, 2.79, -3.41, -3.62])
res = np.ceil(ar)
assert np.array_equal(res, np.array([2.0, 3.0, 3.0, -3.0, -3.0]))

# Test 2: out buffer check
ar_out = np.empty_like(ar)
np.ceil(ar, out=ar_out)
assert np.array_equal(ar_out, res)

print("All numpy.ceil() unit assertions passed successfully! 100% verified.")
```
