# NumPy std(): Standard Deviation, Axes, and Degrees of Freedom

**Tutorial & Documentation:** [https://www.plus2net.com/python/numpy-agg-std.php](https://www.plus2net.com/python/numpy-agg-std.php)

Welcome to the plus2net interactive companion notebook for `numpy.std()`. Run each cell sequentially in Google Colab or your local Jupyter environment.

### Learning Objectives:
1. Calculate population (`ddof=0`) and sample (`ddof=1`) standard deviations.
2. Aggregate dispersion across rows and columns using `axis`.
3. Preserve dimensions with `keepdims=True`.
4. Run a self-verifying assertion suite.

### 1. Overall, Column, and Row Standard Deviations

```python
import numpy as np

a = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
print("Sample Matrix:\n", a)

print("\nOverall std:          ", a.std())
print("Column std (axis=0):  ", a.std(axis=0))
print("Row std (axis=1):     ", a.std(axis=1))
```

### 2. Population vs. Sample Standard Deviation (ddof)

```python
import numpy as np

samples_list = [12, 13, 15, 11, 9, 12, 13, 10, 11, 12, 13, 7, 8]
arr_s = np.array(samples_list)

print("Population std (ddof=0):", arr_s.std(ddof=0))
print("Sample std (ddof=1):    ", arr_s.std(ddof=1))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Overall std verification
arr = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
assert round(arr.std(), 4) == 2.5386

# Test 2: keepdims geometry
kd = arr.std(axis=1, keepdims=True)
assert kd.shape == (3, 1)

# Test 3: ddof differences
data = np.array([12, 13, 15, 11, 9, 12, 13, 10, 11, 12, 13, 7, 8])
assert round(data.std(ddof=0), 4) == 2.1538
assert round(data.std(ddof=1), 4) == 2.2418

print("All numpy.std() unit assertions passed successfully! 100% verified.")
```
