# Creating Array in NumPy
Different methods to create NumPy arrays including `array()`, `eye()`, `ones()`, `arange()`, `linspace()`, `random`, `zeros()`, `full()`, `empty()`, and multidimensional arrays.

```python
import numpy as np
print(np.__version__)
```

## 1. Creating 1D and 2D Arrays

```python
ar1 = np.array([1, 5, 2, 8, 3])
print("1D Array:", ar1)

ar2 = np.array([[1, 5, 2, 8, 3], [8, 5, 3, 11, 1]])
print("2D Array:\n", ar2)
print("Shape:", ar2.shape)
```

## 2. Creating Arrays Using eye(), ones(), zeros(), full(), empty()

```python
print("eye(3, 4):\n", np.eye(3, 4))
print("ones((4, 3)):\n", np.ones((4, 3)))
print("zeros(4):")
print(np.zeros(4, dtype=int))
print("full((1, 3), 5):\n", np.full((1, 3), 5))
print("empty((1, 3)):\n", np.empty((1, 3)))
```

## 3. arange(), linspace(), and Random Arrays

```python
print("arange:", np.arange(1, 5, dtype=np.int8))
print("linspace:", np.linspace(2, 10, 5))
print("random.rand:", np.random.rand(4))
```

## 4. Multidimensional Arrays with ndmin

```python
ar_nd = np.array([1, 2, 3, 4, 5, 6], ndmin=2)
print(ar_nd)
print('Shape:', ar_nd.shape)
print('Dimension:', ar_nd.ndim)
```
