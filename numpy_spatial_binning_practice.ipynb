## Exercise 1: 1D Coordinate Range Binning
Bin longitude coordinates into discrete spatial intervals using `np.searchsorted`.

```python
import numpy as np

lon_bins = np.linspace(-80.0, -70.0, 6)
gps_longitudes = np.array([-79.5, -77.1, -71.2, -80.0, -69.5])

raw_indices = np.searchsorted(lon_bins, gps_longitudes, side='right') - 1
valid_indices = np.clip(raw_indices, 0, len(lon_bins) - 2)

print("Bin Boundaries: ", lon_bins)
print("Valid Binned:   ", valid_indices)

assert np.array_equal(valid_indices, np.array([0, 1, 4, 0, 4]))
```

## Exercise 2: 2D Spatial Grid Indexing (Latitude & Longitude)
Map 2D GPS points into linear 2D grid cell IDs.

```python
def bin_2d_coordinates(lats, lons, lat_edges, lon_edges):
    row_idx = np.clip(np.searchsorted(lat_edges, lats, side='right') - 1, 0, len(lat_edges) - 2)
    col_idx = np.clip(np.searchsorted(lon_edges, lons, side='right') - 1, 0, len(lon_edges) - 2)
    num_cols = len(lon_edges) - 1
    cell_id = row_idx * num_cols + col_idx
    return row_idx, col_idx, cell_id

lat_edges = np.linspace(10.0, 15.0, 6)
lon_edges = np.linspace(70.0, 75.0, 6)

sample_lats = np.array([10.2, 14.8, 12.5])
sample_lons = np.array([70.5, 74.2, 72.1])

rows, cols, cell_ids = bin_2d_coordinates(sample_lats, sample_lons, lat_edges, lon_edges)

print("Row Indices:   ", rows)
print("Col Indices:   ", cols)
print("Linear Cell IDs:", cell_ids)

assert rows[0] == 0 and cols[0] == 0
assert cell_ids[0] == 0
assert cell_ids[1] == 24
```

## Exercise 3: Density Heatmap Aggregation via `np.bincount`
Aggregate 100,000 GPS points into a 10x10 spatial density matrix.

```python
np.random.seed(42)
lats = np.random.uniform(10.0, 20.0, size=100000)
lons = np.random.uniform(70.0, 80.0, size=100000)

lat_edges = np.linspace(10.0, 20.0, 11)
lon_edges = np.linspace(70.0, 80.0, 11)

r_idx = np.clip(np.searchsorted(lat_edges, lats, side='right') - 1, 0, 9)
c_idx = np.clip(np.searchsorted(lon_edges, lons, side='right') - 1, 0, 9)
cell_ids = r_idx * 10 + c_idx

cell_counts = np.bincount(cell_ids, minlength=100)
heatmap_matrix = cell_counts.reshape(10, 10)

print("Heatmap Shape:", heatmap_matrix.shape)
print("Total Points: ", heatmap_matrix.sum())

assert heatmap_matrix.sum() == 100000
assert heatmap_matrix.shape == (10, 10)
```
