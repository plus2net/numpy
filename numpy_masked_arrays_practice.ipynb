## Exercise 1: Masking Error Codes & Invalid Values
Filter out `-999.0` sentinel error codes, `NaN`, and `Inf` values using `np.ma.masked_where` and `np.ma.masked_invalid`.

```python
import numpy as np

# Telemetry array containing NaN, Inf, and sentinel error codes
raw_telemetry = np.array([12.4, np.nan, 15.1, -999.0, np.inf, 13.8, 14.2])

# Step 1: Mask specific sentinel error code (-999.0)
m_sentinel = np.ma.masked_where(raw_telemetry == -999.0, raw_telemetry)

# Step 2: Mask invalid floating point numbers (NaN and Inf)
m_clean = np.ma.masked_invalid(m_sentinel)

print("Raw Telemetry:      ", raw_telemetry)
print("Masked Array View:  ", m_clean)
print("Valid Element Count:", m_clean.count())
print("Masked Mean:        ", m_clean.mean())

# Self-validating assertion checks
assert m_clean.count() == 4
assert np.isclose(m_clean.mean(), 13.875)
```

## Exercise 2: Multi-Condition Masking for Industrial Telemetry
Mask readings if they contain `-999.0` error codes OR fall outside operating limits (`0.0` to `150.0` deg C).

```python
# Industrial boiler temperature logs
boiler_temp = np.array([85.4, 92.1, -999.0, 180.5, -273.15, 88.9, 91.2])

# Mask if error code OR out-of-operating-range
condition = (boiler_temp == -999.0) | (boiler_temp < 0.0) | (boiler_temp > 150.0)
m_boiler = np.ma.masked_where(condition, boiler_temp)

print("Cleaned Boiler Temps:", m_boiler)
print("Operational Average :", m_boiler.mean())

assert m_boiler.count() == 4
assert np.isclose(m_boiler.mean(), 89.4)
```

## Exercise 3: 2D Grid Matrix Multi-Dimensional Aggregation
Compute station-wise and timestep-wise masked means on a multi-sensor grid.

```python
grid_data = np.array([
    [10.5, 12.1, -999.0, 11.8],
    [20.2, -999.0, 22.4, 21.9],
    [15.0, 14.8, 15.2, -999.0]
])

m_grid = np.ma.masked_where(grid_data == -999.0, grid_data)
station_means = m_grid.mean(axis=1)
timestep_means = m_grid.mean(axis=0)

print("Station Means (Row-wise):  ", station_means)
print("Timestep Means (Col-wise): ", timestep_means)

assert np.isclose(station_means[0], 11.466666666666667)
assert np.isclose(timestep_means[2], 18.8)
```

## Exercise 4: Imputing Missing Values with `.filled()`
Replace masked items with computed median values to export clean NumPy arrays.

```python
stream = np.array([50.0, 52.0, -999.0, 54.0, -999.0, 58.0])
m_stream = np.ma.masked_where(stream == -999.0, stream)

valid_median = np.ma.median(m_stream)
imputed_array = m_stream.filled(valid_median)

print("Original Stream:", stream)
print("Imputed Array:  ", imputed_array)

assert imputed_array[2] == 53.0
assert (imputed_array != -999.0).all()
```
