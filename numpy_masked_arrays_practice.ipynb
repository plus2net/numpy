## Step 1: Masking Error Codes & Invalid Values
Run the cell below to filter out `-999.0` error codes and `NaN` values using `np.ma.masked_where` and `np.ma.masked_invalid`.

```python
import numpy as np

# Sample raw telemetry readings with errors
raw_telemetry = np.array([12.4, np.nan, 15.1, -999.0, 13.8])

# Mask -999.0 error codes
m_code = np.ma.masked_where(raw_telemetry == -999.0, raw_telemetry)

# Mask floating-point NaN & Inf values
m_clean = np.ma.masked_invalid(m_code)

print("Raw Telemetry: ", raw_telemetry)
print("Cleaned Masked Array:", m_clean)
print("Valid Element Count: ", m_clean.count())
print("Masked Mean:         ", m_clean.mean())

# Self-validating assertion checks
assert m_clean.count() == 3
assert np.isclose(m_clean.mean(), 13.766666666666667)
```

## Step 2: Imputing Missing Values with `.filled()`
Convert masked arrays back to standard NumPy arrays by replacing masked slots with clean summary metrics like the array median.

```python
incoming_logs = np.array([45.2, 46.1, -999.0, 44.8, 47.0, np.nan, 45.5])

# Dual-pass masking
m_logs = np.ma.masked_where(incoming_logs == -999.0, incoming_logs)
m_logs = np.ma.masked_invalid(m_logs)

computed_median = np.ma.median(m_logs)
imputed_array = m_logs.filled(computed_median)

print("Masked Logs:    ", m_logs)
print("Imputed Array:  ", imputed_array)

assert not np.isnan(imputed_array).any()
assert (imputed_array != -999.0).all()
```
