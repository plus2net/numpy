### Part 1: Ranges & Traversal

### Part 2: Step Increments & Math

```python
# Q4: Evens and countdown
evens = list(range(2, 21, 2))
print("Evens up to 20:", evens)
countdown = list(range(10, 0, -1))
print("Countdown:", countdown)
```

### Part 3: Break, Continue & for...else

```python
# Q9: Prime number check with for...else
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    else:
        return True

print("Is 17 prime?", is_prime(17))
print("Is 20 prime?", is_prime(20))
```

### Part 4: Nested Loops & Transposition

```python
# Q11: Matrix transposition
matrix = [[1, 2], [3, 4], [5, 6]]
transposed = []
for c in range(len(matrix[0])):
    transposed.append([matrix[r][c] for r in range(len(matrix))])
print("Transposed matrix:", transposed)
```
