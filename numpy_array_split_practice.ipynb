### Learning Objectives:
1. Split 1D arrays into equal and unequal sections.
2. Split 2D matrices along `axis=0`.
3. Access nested elements within returned sub-arrays.
4. Run a self-verifying assertion test suite.

### 1. Equal and Unequal Section Splitting

```python
import numpy as np

my_ar = np.arange(9)
print("Split into 3 equal sections:\n", np.array_split(my_ar, 3))
print("\nSplit into 5 unequal sections:\n", np.array_split(my_ar, 5))
```

### 2. Splitting 2D Arrays and Element Access

```python
import numpy as np

my_data = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
my_ar = np.array_split(my_data, 3)

print("Sub-array index 1:", my_ar[1])
print("Row element 0:   ", my_ar[1][0])
print("Value at [1][0][2]:", my_ar[1][0][2])
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Equal sections count and values
arr = np.arange(9)
res3 = np.array_split(arr, 3)
assert len(res3) == 3
assert np.array_equal(res3[0], np.array([0, 1, 2]))

# Test 2: Unequal sections remainder distribution
res5 = np.array_split(arr, 5)
assert len(res5) == 5
assert res5[0].size == 2
assert res5[-1].size == 1

# Test 3: 2D array element indexing
matrix = np.array([[6, 3, 2], [7, 2, 2], [6, 2, 9]])
split_mat = np.array_split(matrix, 3)
assert split_mat[1][0][2] == 2

print("All numpy.array_split() unit assertions passed successfully! 100% verified.")
```
