### Learning Objectives:
1. Convert degree angles to radians using `np.radians()`.
2. Utilize preallocated `out` buffers for performance optimization.
3. Apply boolean `where` masks for selective angle conversions.
4. Run a self-verifying assertion test suite.

### 1. Basic radians() Execution

```python
import numpy as np

ar_deg = np.array([20, 90, 120, 160, 180])
ar_rad = np.radians(ar_deg)
print("Radian result:", ar_rad)
```

### 2. out Buffer and where Mask

```python
import numpy as np

ar_deg = np.array([120, 160, 180])
ar_out = np.ones(ar_deg.shape)
np.radians(ar_deg, out=ar_out)
print(
    "Preallocated out buffer:", ar_out
)

ar_where = np.array([True, False, True])
print("Masked conversion:", np.radians(ar_deg, where=ar_where))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Basic conversion equality
ar_deg = np.array([20, 90, 120, 160, 180])
res = np.radians(ar_deg)
assert round(res[1], 4) == 1.5708

# Test 2: out buffer check and object identity
ar_deg = np.array([120, 160, 180])
ar_out = np.ones(ar_deg.shape)
ans = np.radians(ar_deg, out=ar_out)
assert ans is ar_out

print("All numpy.radians() unit assertions passed successfully! 100% verified.")
```
