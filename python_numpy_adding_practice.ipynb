### 1. Default Flattening vs. Preserving Dimensions

```python
import numpy as np

base_matrix = np.array([[1, 2, 3], [4, 5, 6]])
new_elements = np.array([7, 8, 9])

# Without specifying axis: array is flattened into 1D
flat_result = np.append(base_matrix, new_elements)
print("Flattened result: ", flat_result)
print("Flattened shape:  ", flat_result.shape)
```

### 2. Appending Rows along Axis 0

```python
import numpy as np

npr = np.array([
    [0, 1, 2, 4],
    [3, 4, 5, 6],
    [6, 7, 8, 9]
])

# Must be 2D with matching column width (1 row, 4 cols)
row_to_add = np.array([[10, 11, 12, 13]])

with_row = np.append(npr, row_to_add, axis=0)
print("Appended Row Matrix:\n", with_row)
print("Shape:", with_row.shape)
```

### 3. Appending Columns along Axis 1

```python
import numpy as np

npr = np.array([
    [0, 1, 2, 4],
    [3, 4, 5, 6],
    [6, 7, 8, 9]
])

# 3 rows matching 'npr', any number of columns
cols_to_add = np.array([
    [10, 11],
    [21, 22],
    [31, 32]
])

with_cols = np.append(npr, cols_to_add, axis=1)
print("Appended Columns Matrix:\n", with_cols)
print("Shape:", with_cols.shape)
```

### 4. Resolving Dimension Mismatches with newaxis and reshape

```python
import numpy as np

matrix = np.ones((3, 3), dtype=int)
vector_1d = np.array([5, 5, 5])

# Convert 1D vector (3,) into a 2D row (1, 3)
row_2d = vector_1d[np.newaxis, :]
appended_row = np.append(matrix, row_2d, axis=0)

# Convert 1D vector (3,) into a 2D column (3, 1)
col_2d = vector_1d[:, np.newaxis]
appended_col = np.append(matrix, col_2d, axis=1)

print("Row addition shape:", appended_row.shape)
print("Col addition shape:", appended_col.shape)
```

### 5. Stacking and Concatenation Alternatives

```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

# Vertical stack (axis 0)
v_stacked = np.vstack([a, b])

# Horizontal stack (axis 1)
h_stacked = np.hstack([a, b])

print("vstack:\n", v_stacked)
print("hstack:\n", h_stacked)
```

### 6. Self-Checking Unit Verification Suite
