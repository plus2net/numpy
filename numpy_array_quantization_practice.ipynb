## Exercise 1: Precision Downcasting (float64 -> float32 -> float16)
Downcast arrays to smaller precision types and inspect memory reductions.

```python
import numpy as np

# Sample array in float64
arr_64 = np.array([0.000123456789, 12345.6789, 60000.0], dtype=np.float64)

# Downcast to float32 and float16
arr_32 = arr_64.astype(np.float32)
arr_16 = arr_64.astype(np.float16)

print("float64 Bytes:", arr_64.nbytes)
print("float32 Bytes:", arr_32.nbytes)
print("float16 Bytes:", arr_16.nbytes)

# Self-validating assertions
assert arr_32.nbytes == arr_64.nbytes // 2
assert arr_16.nbytes == arr_64.nbytes // 4
```

## Exercise 2: Linear Min-Max int8 Quantization
Quantize float tensors to int8 and reconstruct them back using scaling parameters.

```python
def quantize_tensor(tensor):
    min_val = tensor.min()
    max_val = tensor.max()
    scale = (max_val - min_val) / 255.0
    zero_point = min_val
    q_tensor = np.round((tensor - zero_point) / scale - 128).astype(np.int8)
    return q_tensor, scale, zero_point

def dequantize_tensor(q_tensor, scale, zero_point):
    return (q_tensor.astype(np.float32) + 128) * scale + zero_point

original_data = np.array([[-15.2], [0.0], [3.14159], [42.8]], dtype=np.float64)
q_data, scale, zero_pt = quantize_tensor(original_data)
reconstructed_data = dequantize_tensor(q_data, scale, zero_pt)

print("Quantized int8 Array:\n", q_data)
print("Reconstructed Data:\n", reconstructed_data)

assert q_data.dtype == np.int8
assert np.allclose(original_data, reconstructed_data, atol=0.5)
```

## Exercise 3: Percentile Clipping for Outlier Handling
Clip extreme signal spikes before quantizing to uint8.

```python
raw_signals = np.array([1.2, 1.5, 1.8, 2.1, 9999.0, 1.4, 1.6], dtype=np.float64)

p_low, p_high = np.percentile(raw_signals, [0, 95])
clipped_signals = np.clip(raw_signals, p_low, p_high)

scale = (p_high - p_low) / 255.0
q_signals = np.round((clipped_signals - p_low) / scale).astype(np.uint8)

print("Clipped Signals: ", clipped_signals)
print("Quantized uint8: ", q_signals)

assert q_signals.max() == 255
assert q_signals.dtype == np.uint8
```
