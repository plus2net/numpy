### Learning Objectives:
1. Convert radian angles to degrees using `np.rad2deg()`.
2. Utilize preallocated `out` buffers for performance optimization.
3. Apply boolean `where` masks for selective angle conversions.
4. Run a self-verifying assertion test suite.

### 1. Basic rad2deg() Execution

```python
import numpy as np

ar_rad = np.array([1.57079633, 2.0943951, 2.7925268, 3.14159265])
ar_deg = np.rad2deg(ar_rad)
print("Degrees result:", ar_deg)
print("Rounded degrees:", np.round(ar_deg))
```

### 2. out Buffer and where Mask

```python
import numpy as np

ar_rad = np.array([2.0943951, 2.7925268, 3.14159265])
ar_out = np.ones(ar_rad.shape)
np.rad2deg(ar_rad, out=ar_out)
print(
    "Preallocated out buffer:", ar_out
)

ar_where = np.array([True, False, True])
print("Masked conversion:", np.rad2deg(ar_rad, where=ar_where))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Basic conversion rounding equality
ar_rad = np.array([1.57079633, 2.0943951, 2.7925268, 3.14159265])
res = np.round(np.rad2deg(ar_rad))
assert np.array_equal(res, np.array([90., 120., 160., 180.]))

# Test 2: out buffer check
ar_out = np.ones(ar_rad.shape)
np.rad2deg(ar_rad, out=ar_out)
assert np.allclose(ar_out, np.rad2deg(ar_rad))

print("All numpy.rad2deg() unit assertions passed successfully! 100% verified.")
```
