### 1. Arithmetic Vectorization

```python
import numpy as np

a = np.array([1.0, 2.0, 3.0, 4.0])
b = np.array([10.0, 20.0, 30.0, 40.0])
res = 2.0 * (a ** 2) + 3.0 * b - 5.0
print("Vectorized result:", res)
```

### 2. Conditional Selection and Meshgrid

```python
import numpy as np

scores = np.array([45, 82, 67, 91, 59])
grades = np.where(scores >= 60, 'Pass', 'Fail')
print("Grades:", grades)

x = np.array([1, 2, 3])
y = np.array([10, 20])
X, Y = np.meshgrid(x, y)
print("Meshgrid Z:\n", X**2 + Y**2)
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Arithmetic vectorization equality
a = np.array([1.0, 2.0])
b = np.array([10.0, 20.0])
assert np.array_equal(2.0 * (a**2) + 3.0 * b - 5.0, np.array([25.0, 63.0]))

# Test 2: np.where conditional
scores = np.array([45, 82])
assert np.array_equal(np.where(scores >= 60, 'Pass', 'Fail'), np.array(['Fail', 'Pass']))

# Test 3: meshgrid shape
X, Y = np.meshgrid([1, 2], [3, 4])
assert X.shape == (2, 2)

print("All numpy vectorization patterns unit assertions passed successfully! 100% verified.")
```
