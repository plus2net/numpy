### 1. Basic 1D and 2D Zero Arrays

```python
import numpy as np

# 1D vector (default float64)
vec = np.zeros(5)
print("1D vector: ", vec)
print("Dtype:     ", vec.dtype)

# 2D matrix (3 rows, 4 cols)
mat = np.zeros((3, 4), dtype=int)
print("\n2D matrix:\n", mat)
```

### 2. Custom Data Types (int, bool, complex)

```python
import numpy as np

z_bool = np.zeros(4, dtype=bool)
z_cplx = np.zeros(3, dtype=complex)
print("Boolean zeros: ", z_bool)
print("Complex zeros: ", z_cplx)
```

### 3. Cloning Geometries with zeros_like()

```python
import numpy as np

prototype = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int16)
mirrored = np.zeros_like(prototype)
print("Prototype shape:", prototype.shape)
print("Mirrored array:\n", mirrored)
```

### 4. Production Recipe: Preallocated Image Mask

```python
import numpy as np

# Create 720p grayscale mask (height=720, width=1280)
mask = np.zeros((720, 1280), dtype=np.uint8)
mask[200:500, 400:800] = 255  # Fill bounding box with white
print("Mask shape:         ", mask.shape)
print("Active white pixels:", np.count_nonzero(mask))
```

### 5. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape integrity
z1 = np.zeros(6)
assert z1.shape == (6,) and np.all(z1 == 0.0)

# Test 2: Multidimensional integer matrix
z2 = np.zeros((2, 3, 2), dtype=int)
assert z2.shape == (2, 3, 2) and z2.size == 12

# Test 3: zeros_like equivalence
source_arr = np.ones((4, 4), dtype=np.float32)
cloned = np.zeros_like(source_arr)
assert cloned.shape == (4, 4) and cloned.dtype == np.float32 and np.all(cloned == 0)

print("All numpy.zeros() unit assertions successfully verified! 100% passed.")
```
