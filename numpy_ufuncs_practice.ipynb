### Learning Objectives:
1. Execute unary and binary ufuncs.
2. Use ufunc methods (`reduce`, `accumulate`, `outer`).
3. Utilize preallocated `out` buffers.
4. Run self-verifying assertion tests.

### 1. Unary and Binary Ufuncs

```python
import numpy as np

x = np.array([1, 4, 9, 16])
print("Square root:", np.sqrt(x))

a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print("Add ufunc:", np.add(a, b))
```

### 2. Ufunc Methods (reduce, accumulate, outer)

```python
import numpy as np

print("Reduce sum:", np.add.reduce([1, 2, 3, 4]))
print("Accumulate sum:", np.add.accumulate([1, 2, 3, 4]))
print("Outer product:\n", np.multiply.outer([1, 2, 3], [1, 2, 3]))
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Unary sqrt
x = np.array([1, 4, 9, 16])
assert np.array_equal(np.sqrt(x), np.array([1., 2., 3., 4.]))

# Test 2: Reduce method
assert np.add.reduce([1, 2, 3, 4]) == 10

# Test 3: Preallocated out buffer
buf = np.empty_like(x, dtype=float)
np.sqrt(x, out=buf)
assert np.array_equal(buf, np.array([1., 2., 3., 4.]))

print("All numpy ufuncs unit assertions passed successfully! 100% verified.")
```
