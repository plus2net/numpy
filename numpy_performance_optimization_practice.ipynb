### 1. Vectorization Speedup

```python
import numpy as np

arr = np.arange(1000000)
result = arr + 10
print("Vectorized result sample:", result[:5])
```

### 2. C-Contiguous Memory and out Buffers

```python
import numpy as np

ar = np.arange(12).reshape(3, 4)
print("Is C-contiguous:", ar.flags['C_CONTIGUOUS'])

a = np.ones(100)
b = np.ones(100)
out_buf = np.empty_like(a)
np.add(a, b, out=out_buf)
print("Preallocated out buffer sum sample:", out_buf[:5])
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Vectorized equality
data = np.arange(1000)
assert np.array_equal(data + 10, data + 10)

# Test 2: C-contiguous flag check
mat = np.arange(6).reshape(2, 3)
assert mat.flags['C_CONTIGUOUS'] == True

# Test 3: Preallocated buffer check
a = np.ones(10)
b = np.ones(10)
buf = np.empty_like(a)
np.add(a, b, out=buf)
assert buf[0] == 2.0

print("All numpy performance optimization unit assertions passed successfully! 100% verified.")
```
