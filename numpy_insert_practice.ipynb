### Learning Objectives:
1. Insert values at specific index positions in 1D arrays.
2. Append elements at the end of arrays using `len(arr)`.
3. Insert rows (`axis=0`) and columns (`axis=1`) in 2D arrays.
4. Run a self-verifying assertion test suite.

### 1. Basic 1D Insertion and Append

```python
import numpy as np

npr = np.array([5, 8, 3])
npr1 = np.insert(npr, 1, 10)
print("Inserted at index 1:", npr1)

npr_end = np.insert(npr, len(npr), 10)
print("Inserted at end:    ", npr_end)
```

### 2. 2D Row and Column Insertion

```python
import numpy as np

npr = np.array([[0, 1, 2, 4], [3, 4, 5, 6], [6, 7, 8, 9]])
row_inserted = np.insert(npr, 1, [10, 11, 12, 13], axis=0)
print("Row inserted matrix:\n", row_inserted)

col_inserted = np.insert(npr, 1, [10, 11, 12], axis=1)
print("\nColumn inserted matrix:\n", col_inserted)
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: 1D insertion equality
arr = np.array([5, 8, 3])
res = np.insert(arr, 1, 10)
assert np.array_equal(res, np.array([5, 10, 8, 3]))

# Test 2: 2D row insertion shape
mat = np.array([[0, 1, 2, 4], [3, 4, 5, 6], [6, 7, 8, 9]])
mat_row = np.insert(mat, 1, [10, 11, 12, 13], axis=0)
assert mat_row.shape == (4, 4)

# Test 3: Flattened fallback without axis
mat_flat = np.insert(mat, 1, [10, 11, 12])
assert mat_flat.ndim == 1

print("All numpy.insert() unit assertions passed successfully! 100% verified.")
```
