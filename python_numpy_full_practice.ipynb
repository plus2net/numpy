### 1. Basic 1D and 2D Constant Arrays

```python
import numpy as np

# 1D vector filled with 7
vec = np.full(5, 7)
print("1D vector: ", vec)
print("Dtype:     ", vec.dtype)

# 2D matrix (3 rows, 4 cols) filled with 3.14
mat = np.full((3, 4), 3.14)
print("\n2D matrix:\n", mat)
```

### 2. Sentinel Values (NaN and Inf)

```python
import numpy as np

nan_mat = np.full((2, 3), np.nan)
inf_mat = np.full((2, 3), np.inf)
print("NaN matrix:\n", nan_mat)
print("Inf matrix:\n", inf_mat)
```

### 3. Cloning Geometries with full_like()

```python
import numpy as np

proto = np.zeros((2, 3), dtype=int)
f_like = np.full_like(proto, 99)
print("Mirrored array:\n", f_like)
```

### 4. Production Recipe: Infinity Cost Matrix

```python
import numpy as np

cost_matrix = np.full((4, 4), np.inf)
np.fill_diagonal(cost_matrix, 0.0)
print("Initial path matrix:\n", cost_matrix)
```

### 5. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape and fill value
f1 = np.full(5, 42)
assert f1.shape == (5,) and np.all(f1 == 42)

# Test 2: NaN grid check
f_nan = np.full((2, 3), np.nan)
assert f_nan.shape == (2, 3) and np.all(np.isnan(f_nan))

# Test 3: full_like equivalence
base = np.zeros((3, 3), dtype=np.int32)
fl = np.full_like(base, 100)
assert fl.shape == (3, 3) and fl.dtype == np.int32 and np.all(fl == 100)

print("All numpy.full() unit assertions successfully verified! 100% passed.")
```
