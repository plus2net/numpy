### 1. Generating 1D and 2D Uniform Floats

```python
import numpy as np

np.random.seed(42)
samples = np.random.random_sample(size=(2, 3))
print("Random sample matrix:\n", samples)
```

### 2. Aliases and Range Scaling

```python
import numpy as np

print("np.random.random(3):", np.random.random(3))
print("np.random.ranf(3):  ", np.random.ranf(3))
print("np.random.sample(3):", np.random.sample(3))

# Scale to [5, 15)
scaled = (15.0 - 5.0) * np.random.random_sample(4) + 5.0
print("Scaled samples:    ", scaled)
```

### 3. Self-Checking Unit Verification Suite

```python
import numpy as np

# Test 1: Shape and interval verification
arr = np.random.random_sample(size=(3, 3))
assert arr.shape == (3, 3)
assert np.all(arr >= 0.0) and np.all(arr < 1.0)

# Test 2: Alias equivalence
np.random.seed(123)
a = np.random.random(4)
np.random.seed(123)
b = np.random.sample(4)
assert np.array_equal(a, b)

# Test 3: Modern generator
rng = np.random.default_rng(42)
mod_arr = rng.random(size=(2, 2))
assert mod_arr.shape == (2, 2)

print("All numpy.random.random_sample() unit assertions passed successfully! 100% verified.")
```
