# NumPy random.randn(): Standard Normal Distribution Practice

### Learning Objectives:
1. Generate scalar, 1D, 2D, and higher-dimensional standard normal variates.
2. Scale samples to an arbitrary normal distribution with mean $\mu$ and standard deviation $\sigma$.
3. Understand dimension argument conventions and avoid common tuple shape errors.
4. Compare legacy `randn()` with the modern `default_rng().normal()` method.
5. Run self-verifying assertion tests.

### 1. Generating Scalars and 1D Vectors

```python
import numpy as np

np.random.seed(42)

# 1. Scalar output (no parameters)
val = np.random.randn()
print("Single float:", val)
print("Type:        ", type(val))

# 2. 1D array of 5 elements
vec = np.random.randn(5)
print("\n1D Vector:", vec)
print("Shape:    ", vec.shape)
```

### 2. Generating 2D Matrices and Multi-dimensional Tensors

```python
import numpy as np

np.random.seed(100)

# 2D matrix (4 rows, 3 columns) - pass as separate arguments, not a tuple
mat = np.random.randn(4, 3)
print("2D Matrix (4x3):\n", mat)

# 3D tensor (2 blocks, 3 rows, 4 columns)
tensor = np.random.randn(2, 3, 4)
print("\n3D Tensor shape:", tensor.shape)
print("Total elements: ", tensor.size)
```

### 3. Scaling to Arbitrary Mean ($\mu$) and Standard Deviation ($\sigma$)

```python
import numpy as np

# Target distribution: Mean = 100, StdDev = 15
mu = 100.0
sigma = 15.0

np.random.seed(7)
iq_scores = mu + sigma * np.random.randn(10000)

print("Sample Mean:  ", np.mean(iq_scores))
print("Sample StdDev:", np.std(iq_scores))
```

### 4. Modern Alternative: default_rng().normal()

```python
import numpy as np

# Modern Generator workflow (NumPy 1.17+)
rng = np.random.default_rng(seed=42)
modern_mat = rng.normal(loc=0.0, scale=1.0, size=(3, 4))

print("Generated with default_rng().normal():\n", modern_mat)
```

### 5. Self-Verifying Unit Test Suite

```python
import numpy as np

# Test 1: Scalar return when no argument is passed
s = np.random.randn()
assert isinstance(s, float)

# Test 2: Shape and dimension integrity
arr_2d = np.random.randn(5, 2)
assert arr_2d.shape == (5, 2) and arr_2d.ndim == 2

# Test 3: Standard deviation and mean properties on large sample
large_sample = np.random.randn(200000)
assert abs(np.mean(large_sample) - 0.0) < 0.02
assert abs(np.std(large_sample) - 1.0) < 0.02

# Test 4: Tuple error handling
try:
    np.random.randn((3, 3)) # Should raise TypeError
    assert False, "Failed to raise TypeError on tuple input"
except TypeError:
    pass

print("All numpy.random.randn() practice assertions passed successfully!")
```
