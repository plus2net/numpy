**Objectives:**
1. Create binary disk arrays using `np.memmap`.
2. Perform chunked reads and in-place writes without OOM errors.
3. Flush changes and verify disk persistence.

```python
import numpy as np

# Create a 2000x2000 float64 memory mapped array
filename = 'test_memmap.dat'
fp = np.memmap(filename, dtype='float64', mode='w+', shape=(2000, 2000))
fp[0, :] = 42.0
fp.flush()
print('Successfully wrote memmap array to disk.')
```

```python
# Read back selectively
fp_read = np.memmap(filename, dtype='float64', mode='r', shape=(2000, 2000))
print('First row sample:', fp_read[0, :5])
assert fp_read[0, 0] == 42.0
```
